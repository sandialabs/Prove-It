```python
import proveit
theory = proveit.Theory() # the theorem's theory
from proveit import defaults, display_provers # useful imports
defaults.assumptions = [] # often useful to set
defaults.auto_simplify = False # can be convenient to enable

from proveit import Lambda, Function
from proveit import a, b, x, y, P, Pxy, Q, Qx
from proveit.logic import Or, And, Not, Iff, Equals
from proveit.core_expr_types import unary_function_existence
from proveit.logic.booleans.quantification.existence import unique_exists_def, unique_existence_by_equality
```

```python
%proving unary_partial_function_existence
```

```python
goal_antecedent = unary_partial_function_existence.instance_expr.antecedent
```

```python
defaults.assumptions = [goal_antecedent]
```

```python
unary_function_existence
```

For each $x$, $Q(x)$ is either true or not true.  If it's true, we'll define $f(x) = y$ such that $P(x, y)$.  Otherwise, we'll define $f(x) = x$ (i.e., the identity) to ensure uniqueness.

```python
_P = Lambda((x, y), Or(And(Qx, Function(P, (x, y))), And(Not(Qx), Equals(y, x))))
```

```python
unary_fn_existence_inst = unary_function_existence.instantiate({P:_P, a:x, b:y})
```

```python
desired_antecedent = unary_fn_existence_inst.antecedent
```

We wish to prove the `desired_antecedent` via `goal_antecedent`.  We do this via a constructive dilemma.  For the constructive dilemma, we first prove the following two statements that have identical consequents:

$Q(x) \Rightarrow \exists!_y \left(\left(Q(x) \land ~P(x, y) \right) \lor \left(\lnot Q(x) \land (y=x) \right)\right)$

$\lnot Q(x) \Rightarrow \exists!_y \left(\left(Q(x) \land ~P(x, y) \right) \lor \left(\lnot Q(x) \land (y=x) \right)\right)$

Start with the first one with $Q(x)$ as the antecedent.

```python
exists_unique_y_st_Pxy = goal_antecedent.instantiate(append_assumptions=[Qx])
```

```python
defaults.assumptions = [Qx, Pxy]
```

```python
y_eq_x = Equals(y, x)
```

```python
Qx_and_Pxy = And(Qx, Pxy).prove()
```

```python
new_existential_instance_for_Qx = Or(Qx_and_Pxy, And(Not(Qx), y_eq_x)).conclude_via_left()
```

```python
exists_unique_y_st_Pxy.substitute_instance(new_existential_instance_for_Qx)
```

Now for the second statement that we need for a constructive dilemma that has $\lnot Q(x)$ as the antecedent.

```python
unique_existence_by_equality
```

```python
exists_unique_y_eq_x = unique_existence_by_equality.instantiate({x:y, y:x})
```

```python
defaults.assumptions = [Not(Qx), y_eq_x]
```

```python
notQx_and__x_eq_y = And(Not(Qx), y_eq_x).prove()
```

```python
new_existential_instance_for_notQx = Or(Qx_and_Pxy, notQx_and__x_eq_y).conclude_via_right()
```

```python
exists_unique_y_eq_x.substitute_instance(new_existential_instance_for_notQx)
```

Now we are ready to prove this unique existence by a constructive dilemma.

```python
Qx_or_notQx = Or(Qx, Not(Qx)).prove()
```

```python
desired_antecedent
```

```python
defaults.assumptions = [goal_antecedent]
```

```python
Qx_or_notQx.derive_via_dilemma(desired_antecedent.instance_expr)
```

We have proven the desired antecedent from the goal antecedent.  Now we can derive the consequent of `unary_fn_existence_inst`:

```python
f_existence_superfluous = unary_fn_existence_inst.derive_consequent()
```

This is almost what we want, but we need to replace the right side with just $P(x, y)$ under the assumption that $Q(x)$ is true.

```python
defaults.assumptions = [goal_antecedent, Qx]
```

```python
superfluous_rhs = f_existence_superfluous.instance_expr.instance_expr.rhs
```

```python
superfluous_rhs.derive_left_if_not_right(append_assumptions=[superfluous_rhs])
```

```python
superfluous_rhs.conclude_via_left(append_assumptions=[Pxy])
```

```python
Iff(superfluous_rhs, Pxy).prove()
```

Now we want to substitute the instance of the `f_existence_superfluous` existential so it matches our goal consequent.

```python
goal_consequent = unary_partial_function_existence.instance_expr.consequent
```

```python
f_existence_superfluous
```

```python
f_existence_superfluous_inside = f_existence_superfluous.instance_expr.instance_expr
```

```python
goal_consequent_inside = goal_consequent.instance_expr.instance_expr
```

```python
f_existence_superfluous.instance_expr.instantiate(assumptions=[f_existence_superfluous.instance_expr])
```

```python
fx_eq_y__iff__Pxy = goal_consequent_inside.prove(assumptions=[Qx, f_existence_superfluous.instance_expr])
```

```python
goal_existence_instance_expr = fx_eq_y__iff__Pxy.generalize((x, y), conditions=[Qx])
```

```python
f_existence_superfluous.substitute_instance(goal_existence_instance_expr)
```
