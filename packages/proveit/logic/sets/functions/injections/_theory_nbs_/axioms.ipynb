```python
import proveit
# Prepare this notebook for defining the axioms of a theory:
%axioms_notebook # Keep this at the top following 'import proveit'.

from proveit import a, b, f, A, B, fa, fb
from proveit.logic import (And, Forall, Equals, NotEquals, 
                           InSet, IsFunction, IsInjection)
```

```python
%begin axioms
```

```python
is_injection_def = Forall(
    (A, B), Forall(
        f, 
        Equals(IsInjection(f, A, B),
               And(IsFunction(f, A, B),
                   Forall((a, b), NotEquals(fa, fb),
                          domain=A, condition=NotEquals(a, b))))
        .with_wrap_after_operator()))
```
