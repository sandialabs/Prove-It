```python
InSet(x, EmptySet).deduce_in_bool()
```

```python
NotInSet(y, EmptySet).conclude()
```

```python
InSet(x, EmptySet).derive_contradiction(assumptions = [InSet(x, EmptySet)])
```

```python
claim_to_deny = A
bad_impl = Implies(claim_to_deny, InSet(x, EmptySet))
InSet(x, EmptySet).deny_via_contradiction(A, assumptions=[bad_impl])
```

```python
from proveit.logic import in_bool, Not
claim_to_affirm = A
bad_impl = Implies(Not(claim_to_affirm), InSet(x, EmptySet))
InSet(x, EmptySet).affirm_via_contradiction(A, assumptions=[bad_impl, in_bool(A)])
```

Perhaps the affirm_via_contradiction() and deny_via_contradiction() methods shown above makes more sense as more general functions rather than EmptySetMembership methods? a method for the EmptySetNonmembership? The methods seems out of place there and counter-intuitive, almost as if we'd want to call such methods on an implication such as $A \Rightarrow (x \in \emptyset)$ instead. We might just want a general function like deny_via_empty_set_contradiction(A, elem=None, …), as a general function defined in the $\text{empty\_set\_membership.py}$ code?

That approach is illustrated below, importing and executing the `affirm_via_empty_set_contradiction()` and `deny_via_empty_set_contradiction()` functions:

```python
claim_to_affirm = A
bad_impl = Implies(Not(claim_to_affirm), InSet(y, EmptySet))
from proveit.logic.sets.empty_set_membership import affirm_via_empty_set_contradiction
affirm_via_empty_set_contradiction(claim_to_affirm, elem = y, assumptions=[bad_impl, in_bool(A)])
```

```python
claim_to_deny = A
bad_impl = Implies(claim_to_deny, InSet(y, EmptySet))
from proveit.logic.sets.empty_set_membership import deny_via_empty_set_contradiction
deny_via_empty_set_contradiction(claim_to_deny, elem = y, assumptions=[bad_impl, in_bool(A)])
```

#### Testing related to `EmptySet.not_equals_side_effects()`

```python
from proveit import x
from proveit.logic import Exists, NotEquals
Exists(x, InSet(x, A)).prove(assumptions=[NotEquals(A, EmptySet)])
```

```python
from proveit import y
from proveit.logic import Exists, NotEquals
Exists(y, InSet(y, A)).prove(assumptions=[NotEquals(A, EmptySet)])
```

```python
from proveit import z
from proveit.logic import Exists, NotEquals
Exists(z, InSet(z, A)).prove(assumptions=[NotEquals(EmptySet, A)])
```
