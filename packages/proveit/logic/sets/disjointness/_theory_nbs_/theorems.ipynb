Theorems (or conjectures) for the theory of <a class="ProveItLink" href="theory.ipynb">proveit.logic.sets.disjointness</a>

```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.
from proveit import a, b, i, n, x, A, B, C, S, T, X, Y, ExprRange, IndexedVar
from proveit.core_expr_types import A_1_to_n, B_1_to_n
from proveit.logic import (
        AllDisjoint, AllDistinct, And, Difference, Disjoint, Distinct, Equals,
        Forall, Implies, NotEquals, Set, Union)
from proveit.logic.sets import EmptySet, Intersect, SubsetEq
from proveit.numbers import one, Interval, Natural, NaturalPos

%begin theorems
```

```python
disjoint_symmetry = Forall((A, B), Equals(Disjoint(A, B), Disjoint(B, A)))
```

```python
disjoint_singleton = Forall(A, Disjoint(A))
```

```python
all_disjoint_singleton_set = Forall(A, AllDisjoint(Set(A)))
```

```python
all_distinct_singleton_set = Forall(x, AllDistinct(Set(x)))
```

```python
disjoint_pair_def_eq = Forall((A, B), Equals(Disjoint(A, B), Equals(Intersect(A, B), EmptySet)))
```

```python
disjoint_pair_unfolding = Forall((A, B), Equals(Intersect(A, B), EmptySet), conditions = [Disjoint(A,B)])
```

```python
disjoint_pair_folding = Forall((A, B), Disjoint(A,B), conditions = [Equals(Intersect(A, B), EmptySet)])
```

```python
nary_disjoint_def = (
    Forall(n,
    Forall((A_1_to_n),
    Equals(Disjoint(A_1_to_n),
           Forall((X, Y), Disjoint(X, Y),
                  conditions = [NotEquals(X, Y)], domain=Set(A_1_to_n)))
    ),
    domain=Natural)
)
```

```python
nary_disjoint_unfolding = (
    Forall(n,
    Forall((A_1_to_n),
    Forall((X, Y), Disjoint(X, Y),
                  conditions = [NotEquals(X, Y)], domain=Set(A_1_to_n)),
    conditions = [Disjoint(A_1_to_n)]),
    domain=Natural)
)
```

```python
nary_disjoint_folding = (
    Forall(n,
    Forall((A_1_to_n),
    Disjoint(A_1_to_n),
    conditions = [Forall((X, Y), Disjoint(X, Y),
                  conditions = [NotEquals(X, Y)], domain=Set(A_1_to_n))]),
    domain=Natural)
)
```

```python
disjoint_eq_unequal_elements = (
    Forall((A, B),
           Equals(Disjoint(A, B), Forall((a, b), NotEquals(a, b), domains=[A, B]))
    )
)
```

```python
disjoint_diff_binary = Forall((A, B), Disjoint(Difference(A, B), B))
```

```python
disjoint_diff = Forall(n,
       Forall((A, B_1_to_n, C),
       And(ExprRange(i, Disjoint(C, IndexedVar(B, i)), one, n)),
       conditions=[Equals(C, Difference(A, Union(B_1_to_n)))]),
       domain=NaturalPos)
```

```python
disjoint_imp_disjoint_diff_left = (
    Forall((A, B, S), Implies(Disjoint(A, B), Disjoint(Difference(A, S), B)))
)
```

```python
disjoint_imp_disjoint_diff_right = (
    Forall((A, B, S), Implies(Disjoint(A, B), Disjoint(A, Difference(B, S))))
)
```

```python
disjoint_imp_disjoint_diffs = (
    Forall((A, B, S, T), Implies(Disjoint(A, B), Disjoint(Difference(A, S), Difference(B, T))))
)
```

```python
disjoint_imp_disjoint_subsets = (
    Forall((A, B, X, Y),
           Implies(Disjoint(A, B), Disjoint(X, Y)),
    conditions=[SubsetEq(X, A), SubsetEq(Y, B)])
)
```
