Through specification, we'll define the set $z$ that is essentially $\{y~|~(y \in S) \land Q(y)\}$.  Then through replacement, we'll prove the existence of a set $w$ for which $\forall_{y \in z} \exists_{x \in w}~x=f(y)$.  Finally, through specification again, we'll define a set $R$ that is essentially $R = \{x~|~x \in w, \exists_{y \in z}~(x=f(y))\}=\{f(y)~|~y \in z\}=\{f(y)~|~(y \in S) \land Q(y)\}$ which is our ultimate goal to define.

```python
specification_eq_inst = specification_eq.instantiate({f:Q, z:S, x:y, y:z})
```

Choose the $z$ such that $z = \{y~|~(y \in S) \land Q(y)\}$.

```python
z_assumption = specification_eq_inst.choose(z)
```

Given set $z$, we can now prove the existence of $w$ such that $\forall_{y \in z}~\exists_{x \in w}~\left(x = f(y)\right)$ via `replacement` and use Skolemization to choose such a $w$.

```python
replacement_inst = replacement.instantiate({x:y, y:x, A:z, B:w})
```

Choose the $w$ such that $\forall_{y \in z} \exists_{x \in w}~x=f(y)$.

```python
w_assumption = replacement_inst.choose(w)
```

```python
x_existence = w_assumption.instantiate(append_assumptions=[InSet(y, z)])
```

```python
x_assumption = x_existence.choose(x)
```

```python
x_eq_fy = x_assumption.derive_right()
```

```python
fy_in_w_under_x_assumption = x_eq_fy.sub_right_side_into(x_assumption.derive_left())
```

```python
fy_in_w = fy_in_w_under_x_assumption.eliminate(x, assumptions=x_existence.assumptions)
```

Now form $R = \{x~|~(x \in w) \land \exists_{y \in z}(x=f(y))\}$.

```python
specification_eq
```

```python
y_existence = Exists(y, Equals(x, fy), condition=InSet(y, z))
```

```python
R_existence = specification_eq.instantiate({y:R, z:w, fx:y_existence})
```

```python
R_assumption = R_existence.choose(R)
```

We want to eliminate $w$ from the right side above.  We do this by proving $\exists_{y \in z} (x = f(y)) \Rightarrow (x \in w)$ so then $(x \in w)$ is superfluous.

```python
y_existence = R_assumption.instance_expr.rhs.operands[1]
```

```python
y_assumptions = y_existence.choose(y)
```

```python
x_eq_fy = y_assumptions.derive_right()
```

```python
fy_in_w
```

```python
x_in_w__under_y_assumptions = x_eq_fy.sub_left_side_into(fy_in_w, assumptions=x_eq_fy.assumptions+[w_assumption])
```

```python
x_in_w = x_in_w__under_y_assumptions.eliminate(y, assumptions=[y_existence])
```

```python
x_in_w.as_implication(y_existence)
```

```python
R_assumption_inst = R_assumption.instantiate()
```

```python
Implies(y_existence, R_assumption_inst.rhs).prove(assumptions=[w_assumption])
```

```python
R_membership = Equals(R_assumption_inst.lhs, y_existence).prove(assumptions=[R_assumption, w_assumption])
```

```python
simplified_R_existence__under_R_and_w_assumptions = Exists(R, R_membership).conclude_via_example(R, assumptions=[R_assumption, w_assumption])
```

```python
simplified_R_existence__under_w_assumption = simplified_R_existence__under_R_and_w_assumptions.eliminate(R)
```

```python
simplified_R_existence = simplified_R_existence__under_w_assumption.eliminate(w)
```
