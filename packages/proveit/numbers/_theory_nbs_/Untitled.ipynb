```python
from proveit import Variable, alpha, beta, defaults, p
from proveit.numbers import Real, Add, Mult, subtract, frac, one, two
from proveit.logic.sets import InSet
from proveit.logic import Equals
```

```python
%begin demo
```

```python
n1, n2, n3, n4 = [Variable('n_{%d}'%i) for i in range(1, 5)]
```

```python
q1, q2, q3, q4 = [Variable('q_{%d}'%i) for i in range(1, 5)]
```

```python
defaults.assumptions = [InSet(n1, Real), InSet(n2, Real), InSet(n3, Real), InSet(n4, Real),
                        InSet(q1, Real), InSet(q2, Real), InSet(q3, Real), InSet(q4, Real), InSet(p, Real),
                        InSet(alpha, Real), InSet(beta, Real)]
```

```python
n12 = Add(n1, n2)
```

```python
Q12_ = subtract(Mult(Add(beta, p), n2), Mult(Add(alpha, p), n1))
```

```python
Equals(Mult(n2, Add(one, frac(Add(beta, p), Add(alpha, p)))), Add(n12, frac(Q12, Add(alpha, p))))
```

```python
Q12 = Variable('Q_{12}')
n12 = Variable('n_{12}')
Q34 = Variable('Q_{34}')
n34 = Variable('n_{34}')
```

```python
Equals(Mult(n2, Add(Add(alpha, p), Add(beta, p))), Add(Mult(Add(alpha, p), n12), Q12))
```

```python
Equals(Mult(n1, Add(Add(beta, p), Add(alpha, p))), subtract(Mult(Add(beta, p), n12), Q12))
```

```python
Equals(Add(alpha, beta), one)
```

```python
Equals(Mult(n2, Add(one, Mult(two, p))), Add(Mult(Add(alpha, p), n12), Q12))
```

```python
Equals(Mult(n1, Add(one, Mult(two, p))), subtract(Mult(Add(beta, p), n12), Q12))
```
