Questions
1)How to distinguish between x and c if we assign them to the same space? We have to assure x is a variable and c is a constant.

```python
diff_x = Forall(y, Equals(Function(Differentiate(Lambda(x,x)),y),one),domain=Real )
```

```python
diff_const = Forall(y, Forall(c, Equals(Function(Differentiate(Lambda(x,c)),y),zero)),domain=Real )
```

```python
diff_neg = Forall(U,
       Forall(f,
              Forall(y,
                    Equals(Function(Differentiate(Lambda(x,Neg(fx))),y),
                           Mult(Neg(one),Function(Differentiate(f),y))),
                domain=U),
        condition = IsDifferentiable(f,U)),
       condition=SubsetEq(U, Real))
```

### Differentiation rules where Lambda maps are defined inside a Function

```python
diff_add = Forall(U,
       Forall((f, g),
              Forall(y,
                    Equals(Function(Differentiate(Lambda(x,Add(fx,gx))),y),
                           Add(Function(Differentiate(f),y),Function(Differentiate(g),y)))),
      conditions = [IsDifferentiable(f,U),IsDifferentiable(g,U)]),
       condition=SubsetEq(U, Real))
```

```python
diff_prod = Forall(U,
       Forall((f, g),
              Forall(y,
                    Equals(Function(Differentiate(Lambda(x,Mult(fx,gx))),y),
                           Add(Mult(Function(Differentiate(f),y),Function(g,y)),Mult(Function(f,y),Function(Differentiate(g),y))))),
       conditions = [IsDifferentiable(f,U),IsDifferentiable(g,U)]),
       condition=SubsetEq(U, Real))
```

For IsDifferentiables

```python
x_isDiff = Forall(U, IsDifferentiable(Lambda(x,x),U),condition=SubsetEq(U, Real))
```

```python
const_isDiff = Forall(U, Forall(c, IsDifferentiable(Lambda(x,c),U)), condition=SubsetEq(U, Real))
```

```python
add_isDiff =  Forall(U,
       Forall((f, g),
                    IsDifferentiable(Lambda(x, Add(fx,gx)),U),
      conditions = [IsDifferentiable(f,U),IsDifferentiable(g,U)]),
       condition=SubsetEq(U, Real))
```

```python
prod_isDiff =  Forall(U,
       Forall((f, g),
                    IsDifferentiable(Lambda(x, Mult(fx,gx)),U),
      conditions = [IsDifferentiable(f,U),IsDifferentiable(g,U)]),
       condition=SubsetEq(U, Real))
```

```python
Neg_isDiff = Forall(U,
       Forall(f, IsDifferentiable(Lambda(x, Neg(fx)),U),
      condition = IsDifferentiable(f,U)),
       condition=SubsetEq(U, Real))
```

```python
exp_natpos_isDiff = Forall(U,
       Forall(f,
              Forall(n, IsDifferentiable(Lambda(x,Exp(fx,n)),U),domain=NaturalPos),
      conditions = [IsDifferentiable(f,U), IsFunction(f,U, Real)]),
       condition=SubsetEq(U, Real))

```

```python
exp_integer_isDiff = Forall(U,
       Forall(f,
              Forall(n, IsDifferentiable(Lambda(x,Exp(fx,n)),U),domain=Integer),
      conditions = [IsDifferentiable(f,U), IsFunction(f,U, RealNonZero)]),
       condition=SubsetEq(U, Real))
```

```python
exp_real_isDiff = Forall(U,
       Forall(f,
              Forall(n, IsDifferentiable(Lambda(x,Exp(fx,n)),U),domain=Real),
      conditions = [IsDifferentiable(f,U), IsFunction(f,U, RealPos)]),
       condition=SubsetEq(U, Real))
```
